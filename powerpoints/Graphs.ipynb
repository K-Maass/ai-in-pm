# Stakeholder Analysis
The following matrix plots our key stakeholders based on their current **Support** and **Impact** on the project. 

* **Top Right (Promoters):** Key gatekeepers we must rely on to push the project forward.
* **Top Left (Manage Closely):** High-impact skeptics whose buy-in is critical for successful implementation.
* **Center (Neutral/Conditional):** Stakeholders who require tailored proofs of concept and day-to-day engagement to shift into promoters.

The visual below maps out the exact needs and current status of each individual to guide our engagement strategy.

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import matplotlib.patheffects as PathEffects
import textwrap

# 1. Setup the figure layout 
fig, ax = plt.subplots(figsize=(15, 9), dpi=150)
fig.patch.set_facecolor('#ffffff')
ax.set_facecolor('#ffffff')

# Disattiviamo gli assi standard e allarghiamo i limiti per fare spazio al testo e agli assi personalizzati
ax.set_xlim(-1.5, 11)
ax.set_ylim(-1.5, 11)
ax.set_axis_off() 

# 2. Quadrant Background Colors (Pastel & Light)
ax.add_patch(patches.Rectangle((0, 5), 5, 5, color='#fff4e6', zorder=0)) # Manage Closely (Orange)
ax.add_patch(patches.Rectangle((5, 5), 5, 5, color='#eafaf1', zorder=0)) # Key Promoters (Green)
ax.add_patch(patches.Rectangle((0, 0), 5, 5, color='#f8f9f9', zorder=0)) # Monitor (Gray)
ax.add_patch(patches.Rectangle((5, 0), 5, 5, color='#ebf5fb', zorder=0)) # Keep Informed (Blue)

# 3. Central Dividers
ax.plot([0, 10], [5, 5], color='#bbbbbb', linestyle='--', linewidth=1.5, zorder=1)
ax.plot([5, 5], [0, 10], color='#bbbbbb', linestyle='--', linewidth=1.5, zorder=1)

# 4. Custom Axes with Arrows (Come da immagine modello)
# X-Axis Line
ax.annotate('', xy=(10.6, 0), xytext=(0, 0),
            arrowprops=dict(arrowstyle="-|>", lw=3, color='#333333', mutation_scale=25), zorder=2)
# Y-Axis Line
ax.annotate('', xy=(0, 10.6), xytext=(0, 0),
            arrowprops=dict(arrowstyle="-|>", lw=3, color='#333333', mutation_scale=25), zorder=2)

# X-Axis Labels (Parola in mezzo, Low e High agli estremi)
ax.text(5, -0.6, "Support", fontsize=15, weight='bold', color='#333333', ha='center', va='top')
ax.text(0, -0.6, "Low", fontsize=13, weight='bold', color='#333333', ha='center', va='top')
ax.text(10, -0.6, "High", fontsize=13, weight='bold', color='#333333', ha='center', va='top')

# Y-Axis Labels (Parola in mezzo, Low e High agli estremi)
ax.text(-0.6, 5, "Impact", fontsize=15, weight='bold', color='#333333', ha='right', va='center')
ax.text(-0.6, 0, "Low", fontsize=13, weight='bold', color='#333333', ha='right', va='center')
ax.text(-0.6, 10, "High", fontsize=13, weight='bold', color='#333333', ha='right', va='center')

# 5. Quadrant Labels in the corners
ax.text(0.2, 9.8, "Manage Closely", color='#d35400', weight='bold', fontsize=14, va='top', ha='left', zorder=2)
ax.text(9.8, 9.8, "Key Promoters", color='#27ae60', weight='bold', fontsize=14, va='top', ha='right', zorder=2)
ax.text(0.2, 0.2, "Monitor", color='#7f8c8d', weight='bold', fontsize=14, va='bottom', ha='left', zorder=2)
ax.text(9.8, 0.2, "Keep Informed", color='#2980b9', weight='bold', fontsize=14, va='bottom', ha='right', zorder=2)

# 6. Define Stakeholder Data 
stakeholders =[
    {
        "name": "Beat Glarner (CEO)", 
        "desc": "Needs visible KPI improvement and a convincing business case",
        "x": 8.5, "y": 8.5, "color": "#d62728"
    },
    {
        "name": "Lauren Hartford (Group CIO)", 
        "desc": "Needs secure architecture, realistic scope, and proof that this is more than AI hype",
        "x": 5.0, "y": 8.5, "color": "#1f77b4"
    },
    {
        "name": "Urs Mueller (Shift Lead)", 
        "desc": "Main skeptic, but also key source of tacit process knowledge",
        "x": 1.5, "y": 8.5, "color": "#ff7f0e" # Spostato più a sinistra (era 2.5)
    },
    {
        "name": "Dave Liebknecht (IT intern)", 
        "desc": "Essential technical bridge, but current single point of failure",
        "x": 8.5, "y": 5.0, "color": "#9467bd" # Abbassato sulla linea di mezzo (era 6.5)
    },
    {
        "name": "Quality Lab / Operators", 
        "desc": "Provide labels, context, and day-to-day adoption",
        "x": 5.0, "y": 3.5, "color": "#2ca02c"
    }
]

# 7. Plot points and text 
for s in stakeholders:
    # Marker
    ax.scatter(s['x'], s['y'], color=s['color'], s=320, zorder=5, edgecolors='black', linewidths=1.5)
    
    # Nome: Distanziato dal pallino, in grassetto, centrato
    txt_name = ax.text(s['x'], s['y'] - 0.35, s['name'], weight='bold', fontsize=11.5, 
                       ha='center', va='top', color='#111111', zorder=6)
    txt_name.set_path_effects([PathEffects.withStroke(linewidth=3.5, foreground='w')])
    
    # Descrizione: Aggiunto più spazio dal nome (ora è y - 0.85 invece di 0.65)
    wrapped_desc = textwrap.fill(s['desc'], width=38)
    txt_desc = ax.text(s['x'], s['y'] - 0.85, wrapped_desc, fontsize=10.5, 
                       ha='center', va='top', color='#333333', linespacing=1.4, zorder=6)
    txt_desc.set_path_effects([PathEffects.withStroke(linewidth=3.5, foreground='w')])

# 8. Title formatting
plt.suptitle("Stakeholder Map: Support vs. Impact", fontsize=22, weight='bold', color='#111111', y=0.98)
plt.title("The project succeeds only if the key gatekeepers buy in", fontsize=15, style='italic', color='#555555', pad=15)

# Render
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import textwrap

# 1. Setup della figura (Formato Slide 16:9)
fig, ax = plt.subplots(figsize=(16, 9), dpi=150)
fig.patch.set_facecolor('#ffffff')
ax.set_facecolor('#ffffff')

# Disattiviamo gli assi per usare la finestra come una tela libera (coordinate da 0 a 1)
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.set_axis_off()

# ==========================================
# SEZIONE SUPERIORE: TESTO DESCRITTIVO
# ==========================================

header_data =[
    {
        "title": "Problem:", 
        "text": "the current sorter setup relies on default logic and manual tuning of only 3 active features, which is failing under this year's summer swell conditions",
        "y": 0.90
    },
    {
        "title": "Current impact:", 
        "text": "contamination escapes, 15% pods/stems in one batch, 20 tons of good peas rejected, yield down to 65%",
        "y": 0.76
    },
    {
        "title": "Objective:", 
        "text": "build a decision-support pilot that links machine data, quality outcomes, and shift context to recommend better sorting settings",
        "y": 0.62
    }
]

# Rendering del testo con layout "a tabella"
for item in header_data:
    # Colonna di sinistra: Prefisso in grassetto
    ax.text(0.04, item['y'], item['title'], fontsize=18, weight='bold', color='#111111', va='top')
    
    # Colonna di destra: Testo formattato a capo
    wrapped_text = textwrap.fill(item['text'], width=88)
    ax.text(0.20, item['y'], wrapped_text, fontsize=18, color='#111111', va='top', linespacing=1.4)


# ==========================================
# SEZIONE INFERIORE: KPI CARDS
# ==========================================

kpi_data =[
    {
        'color': '#1d4e89', # Dark Blue
        'title': 'KPI 01 - Yield',
        'sub1': 'Good-Product Yield Rate',
        'sub2': 'Share of good units out of total inspected • per shift',
        'cur_label': 'Current',
        'cur_val': '65%',
        'tgt_label': 'Target',
        'tgt_val': r'$\geq$ 97%',
        'tgt_size': 32,
        'footer': 'Baseline: 97% • Source: Charge_ID Master Table'
    },
    {
        'color': '#d35400', # Dark Orange
        'title': 'KPI 02 - Contamination',
        'sub1': 'Pods & Stems Rate',
        'sub2': 'Share of contaminated units out of passed • per batch',
        'cur_label': 'Current',
        'cur_val': '15%',
        'tgt_label': 'Target',
        'tgt_val': '< 5%',
        'tgt_size': 32,
        'footer': 'Zero-tolerance: Category A (stones) • Source: Manual_Label'
    },
    {
        'color': '#1e8449', # Dark Green
        'title': 'KPI 03 - Economics',
        'sub1': 'False-Negative Cost / Hour',
        'sub2': 'Cost of quality escapes at 1.5 m/s belt speed • per hour',
        'cur_label': 'Unit cost (FN)',
        'cur_val': '500 CHF',
        'tgt_label': 'Target',
        'tgt_val': 'Minimize',
        'tgt_size': 28, # Leggermente più piccolo per far stare la parola intera
        'footer': 'Weights: FN=500 CHF, FP=10 CHF • Source: Exercise 6'
    }
]

# Parametri geometrici per le card
card_y = 0.06
card_w = 0.28
card_h = 0.46
x_positions =[0.04, 0.36, 0.68]

# Disegno iterativo delle KPI Cards
for x, data in zip(x_positions, kpi_data):
    
    # Sfondo della card e bordo esterno
    ax.add_patch(patches.Rectangle((x, card_y), card_w, card_h, facecolor='#fafbfa', edgecolor='#222222', lw=1.2))
    
    # Striscia colorata sul lato sinistro
    ax.add_patch(patches.Rectangle((x, card_y), 0.008, card_h, facecolor=data['color'], edgecolor='none', zorder=2))
    
    # Helper per il calcolo delle coordinate Y interne
    def y_offset(offset):
        return card_y + card_h - offset

    # Testi superiori della card
    ax.text(x + 0.025, y_offset(0.06), data['title'], fontsize=18, weight='bold', color='#111111')
    ax.text(x + 0.025, y_offset(0.13), data['sub1'], fontsize=14, weight='bold', color='#111111')
    ax.text(x + 0.025, y_offset(0.18), data['sub2'], fontsize=9.5, color='#333333')
    
    # Linea separatrice interna
    ax.plot([x + 0.02, x + card_w - 0.02], [y_offset(0.23), y_offset(0.23)], color='#cccccc', lw=1)
    
    # Metrica: CURRENT (Colonna Sinistra)
    ax.text(x + 0.025, y_offset(0.29), data['cur_label'], fontsize=11, color='#111111')
    ax.text(x + 0.025, y_offset(0.39), data['cur_val'], fontsize=32, weight='bold', color='#c0392b') # Rosso
    
    # Metrica: TARGET (Colonna Destra)
    ax.text(x + 0.15, y_offset(0.29), data['tgt_label'], fontsize=11, color='#111111')
    ax.text(x + 0.15, y_offset(0.39), data['tgt_val'], fontsize=data['tgt_size'], weight='bold', color='#27ae60') # Verde
    
    # Footer
    ax.text(x + 0.025, card_y + 0.025, data['footer'], fontsize=8.5, color='#444444')

# Rendering finale
plt.tight_layout()
plt.show()
```
